## Objective

The objective of this project is to analyze customer shopping data and identify patterns in purchasing behavior, sales, product preferences, payment methods, purchase frequency, and customer reviews.

## Dataset Description

The dataset used for this project is the Customer Shopping Trends dataset obtained from Kaggle.

The dataset contains 3,900 customer records and 18 attributes related to customer shopping behavior. It includes information about customer demographics, products purchased, purchase amount, location, season, review ratings, subscription status, payment methods, shipping type, discounts, previous purchases, and purchase frequency.

The dataset is used to identify customer purchasing patterns and analyze sales-related trends using Python.

## Exploratory Data Analysis

The dataset is analyzed based on product category, purchased items, location, payment method, subscription status, discounts, promotional codes, purchase frequency, and customer ratings.

```python
# 1. Gender analysis
print("GENDER ANALYSIS")
print(df2.groupby("Gender")["Purchase Amount (USD)"].agg(["count", "mean", "sum"]))

# 2. Category analysis
print("\nCATEGORY ANALYSIS")
print(df2.groupby("Category")["Purchase Amount (USD)"].agg(["count", "mean", "sum"]))

# 3. Item analysis
print("\nTOP ITEMS BY PURCHASE COUNT")
print(df2.groupby("Item Purchased")["Purchase Amount (USD)"].agg(["count", "mean", "sum"])
      .sort_values("count", ascending=False).head(10))

# 4. Location analysis
print("\nTOP 10 LOCATIONS BY SALES")
print(df2.groupby("Location")["Purchase Amount (USD)"].agg(["count", "mean", "sum"])
      .sort_values("sum", ascending=False).head(10))

# 5. Payment method analysis
print("\nPAYMENT METHOD ANALYSIS")
print(df2.groupby("Payment Method")["Purchase Amount (USD)"].agg(["count", "mean", "sum"]))

# 6. Subscription analysis
print("\nSUBSCRIPTION ANALYSIS")
print(df2.groupby("Subscription Status")["Purchase Amount (USD)"].agg(["count", "mean", "sum"]))

# 7. Discount analysis
print("\nDISCOUNT ANALYSIS")
print(df2.groupby("Discount Applied")["Purchase Amount (USD)"].agg(["count", "mean", "sum"]))

# 8. Promo code analysis
print("\nPROMO CODE ANALYSIS")
print(df2.groupby("Promo Code Used")["Purchase Amount (USD)"].agg(["count", "mean", "sum"]))

# 9. Purchase frequency analysis
print("\nPURCHASE FREQUENCY ANALYSIS")
print(df2.groupby("Frequency of Purchases")["Purchase Amount (USD)"].agg(["count", "mean", "sum"]))

# 10. Average rating by category
print("\nRATING BY CATEGORY")
print(df2.groupby("Category")["Review Rating"].mean().sort_values(ascending=False))
```

## Data Visualization

The following visualizations are used to represent important patterns and trends in customer purchasing behavior.

```python
# VISUALIZATION

# 1. Purchases by Gender
plt.figure(figsize=(6,4))
sns.countplot(data=df2, x="Gender")
plt.title("Number of Purchases by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Purchases")
plt.show()


# 2. Sales by Category
plt.figure(figsize=(7,4))
category_sales = df2.groupby("Category")["Purchase Amount (USD)"].sum()
category_sales.plot(kind="bar")
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales (USD)")
plt.xticks(rotation=0)
plt.show()


# 3. Top 10 Items by Purchase Count
plt.figure(figsize=(9,5))
top_items = df2["Item Purchased"].value_counts().head(10)
top_items.plot(kind="bar")
plt.title("Top 10 Most Purchased Items")
plt.xlabel("Item")
plt.ylabel("Number of Purchases")
plt.xticks(rotation=45)
plt.show()


# 4. Sales by Payment Method
plt.figure(figsize=(8,4))
payment_sales = df2.groupby("Payment Method")["Purchase Amount (USD)"].sum()
payment_sales.plot(kind="bar")
plt.title("Total Sales by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Sales (USD)")
plt.xticks(rotation=45)
plt.show()


# 5. Sales by Purchase Frequency
plt.figure(figsize=(9,4))
frequency_sales = df2.groupby("Frequency of Purchases")["Purchase Amount (USD)"].sum()
frequency_sales.plot(kind="bar")
plt.title("Total Sales by Purchase Frequency")
plt.xlabel("Purchase Frequency")
plt.ylabel("Total Sales (USD)")
plt.xticks(rotation=45)
plt.show()


# 6. Average Rating by Category
plt.figure(figsize=(7,4))
rating = df2.groupby("Category")["Review Rating"].mean()
rating.plot(kind="bar")
plt.title("Average Review Rating by Category")
plt.xlabel("Category")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.show()


# 7. Purchase Amount Distribution
plt.figure(figsize=(7,4))
plt.hist(df2["Purchase Amount (USD)"], bins=15)
plt.title("Distribution of Purchase Amount")
plt.xlabel("Purchase Amount (USD)")
plt.ylabel("Number of Customers")
plt.show()
```

# KEY FINDINGS

1. Male customers made more purchases than female customers.

2. Clothing generated the highest total sales among all categories.

3. Blouse, Jewelry, and Pants were among the most frequently purchased items.

4. Sales were distributed fairly evenly across the different payment methods.

5. Purchase frequency categories showed relatively similar total sales.

6. Footwear had the highest average review rating among the four categories.

7. Purchase amounts ranged from $20 to $100, showing a wide distribution of spending.
